## 1.1. Методи роботи з рядками

Strings can be transformed using **methods**. Recall that methods and functions are not the same thing. Here is the textbook section on string methods: [4.2.1 String Methods](https://inferentialthinking.com/chapters/04/2/1/String_Methods.html). 

Here's a sketch of how to call methods on a string:

    <expression that evaluates to a string>.<method name>(<argument>, <argument>, ...)
    
One example of a string method is `replace`, which replaces all instances of some part of the original string (or a *substring*) with a new string. 

    <original string>.replace(<old substring>, <new substring>)
    
`replace` returns (evaluates to) a new string, leaving the original string unchanged.
    
Try to predict the output of this example, then run the cell!

Рядки можна змінювати певним чином за допомогою **методів** (**methods**). Нагадаємо, що методи та функції — це не одне й те саме. Ось розділ підручника, присвячений методам роботи з рядками: [4.2.1 Методи роботи з рядками](https://inferentialthinking.com/chapters/04/2/1/String_Methods.html).

Ось схема виклику методів для рядка:

    <вираз, що має значення у вигляді рядка>.<назва методу>(<аргумент>, <аргумент>, ...)

Одним із прикладів методу для рядків є `replace`, який замінює всі входження певної частини вихідного рядка (цю частину називають *підрядок*, *substring*) на новий рядок.

    <оригінальний рядок>.replace(<старий підрядок>, <новий підрядок>)

`replace` повертає (результатом обчислення є) новий рядок, залишаючи оригінальний рядок без змін.

Спробуйте передбачити результат цього прикладу, а потім запустіть комірку!

```python
# Наступний код замінить в рядку 'bag' одну букву (підрядок) на іншу (новий підрядок)
bag = 'bag'
print(bag.replace('g', 't'), bag)
```

Ви можете викликати функції, застосовуючи їх до результатів інших функцій. Наприклад, `max(abs(-5), abs(3))` дає у результаті 5. Аналогічно, ви можете викликати методи, застосовуючи їх до результатів викликів інших методів або функцій.

Можливо, ви вже помітили одну відмінність між функціями та методами — перед викликом функції, такої як `max`, не потрібно ставити `.`, на відміну від методу для рядків, такого як `replace`. Довідник Python, про який йшлося у вступному абзаці, може стати в нагоді, якщо ви не впевнені, як правильно викликати функцію чи метод.

```python
# Використання методу replace на результаті іншого виклику replace
# Послідовність дій є такою: 
# 1) спочатку в рядку 'train' замінюємо 't' на 'ing', ця дія повертає рядок 'ingrain' 
# 2) в рядку 'ingrain' замінюємо 'in' на 'de'
'train'.replace('t', 'ing').replace('in', 'de')
```

# 2. Arrays

Computers are most useful when you can use a small amount of code to *do the same action* to *many different things*.

For example, in the time it takes you to calculate the 18% tip on a restaurant bill, a laptop can calculate 18% tips for every restaurant bill paid by every human on Earth that day.  (That's if you're pretty fast at doing arithmetic in your head!)

**Arrays** are how we put many values in one place so that we can operate on them as a group. For example, if `billions_of_numbers` is an array of numbers, the expression

    .18 * billions_of_numbers

gives a new array of numbers that contains the result of multiplying **each number** in `billions_of_numbers` by .18.  Arrays are not limited to numbers; we can also put all the words in a book into an array of strings.

Concretely, an array is a **collection of values of the same type**.

###  `np.arange`
Масиви в Python забезпечуються пакетом під назвою [NumPy](http://www.numpy.org/) (вимовляється як «NUM-pie»). Пакет називається `numpy`, але зазвичай його перейменовують на `np` для стислості.  Це можна зробити так:

    import numpy as np

У науці про дані дуже часто доводиться працювати з великою кількістю чисел, рівномірно розподілених у певному діапазоні.  NumPy надає для цього спеціальну функцію під назвою `arange`.  Рядок коду `np.arange(start, stop, step)` створює масив, що містить усі числа, починаючи з `start` і збільшуючись на `step`, зупиняючись **до** досягнення `stop`.

Виконайте наведені нижче комірки, щоб побачити кілька прикладів!

```python
# Цей масив починається з 1 (start) і збільшується на 2 (step)
# а потім зупиняється перед 6 (stop)
np.arange(1, 6, 2)
```

```python
# Цей масив не містить числа 9,
# оскільки функція np.arange зупиняється *до* досягнення значення stop
np.arange(4, 9, 1)
```

Це дивна традиція в інформатиці.  **0 називається *індексом* першого елемента.**  Цей індекс елемента показує кількість елементів, що знаходяться *перед* цим елементом.  Отже, 3 — це індекс четвертого елемента (перед ним знаходяться 3 елемнети).

## 2.3. Виконання дії над кожним елементом масиву
Масиви в першу чергу корисні для багатократного виконання однієї й тієї ж операції над усіма елементами, завдяки цьому нам не доводиться часто використовувати `.item` і працювати з окремими елементами.

### Округлення (Rounding)
Ось одне просте запитання, яке ми можемо поставити щодо чисельності населення світу:

> Якою була чисельність населення кожного року, округлена до найближчого мільйона?

Округлення часто використовується з великими числами, коли нам не потрібна надто висока точність у цифрах. Одним із прикладів цього є представлення даних у таблицях та візуалізаціях.

Ми можемо спробувати відповісти на наше запитання, використовуючи вбудовану в Python функцію `round` та метод `item`, який ви щойно побачили.

**Примітка:** функція `round` приймає два аргументи: число, яке потрібно округлити, та кількість знаків після коми, до якої слід округлити. Другий аргумент можна розглядати як кількість кроків вправо або вліво від десяткової коми. Від’ємні числа означають переміщення вліво, а додатні — вправо. Отже, якщо ми маємо `round(1234.5, -2)`, це означає, що слід переміститися на два знаки вліво, а потім замінити всі цифри праворуч від цього знака на нулі. У результаті отримаємо число 1200.0. Такий самий результат ми отримаємо і в цьому випадку `round(1234, -2)` (кома хоч вона тут і не написана явно, розташована після цифри `4`, тобто `1234.0`) З іншого боку, якщо ми маємо `round(6.789, 1)`, нам слід переміститися на одну позицію вправо, а потім замінити всі цифри праворуч від цієї позиції на нулі. У результаті буде виведено число 6.8.


<img src="np_round_example.png">

```python
population_1950_magnitude = round(population_amounts.item(0), -6)
population_1951_magnitude = round(population_amounts.item(1), -6)
population_1952_magnitude = round(population_amounts.item(2), -6)
population_1953_magnitude = round(population_amounts.item(3), -6)
print(population_1950_magnitude, population_1951_magnitude, population_1952_magnitude, population_1953_magnitude)
```

Але це нудно і не дозволяє повною мірою скористатися тим, що ми працюємо з комп’ютером.

Натомість NumPy надає власну версію функції `round`, яка округлює кожен елемент масиву. Вона приймає два аргументи: один - масив чисел та другий - кількість знаків після коми, до якої слід округлювати. Вона повертає масив тієї самої довжини, де перший елемент результату — це округлений перший елемент аргументу, і так далі.

## 3. Створення таблиць

Масив зручно використовувати для опису одного параметра кожного елемента в колекції. Наприклад, припустимо, що наша колекція складається з усіх штатів США. Тоді масив міг би описувати площу кожного штату (один з можливих параметрів).

Таблиці розширюють цю ідею, містячи кілька стовпців (columns), що зберігаються та представлені у вигляді масивів, кожен з яких описує окремий атрибут (параметр) для кожного елемента колекції. Таким чином, таблиці дозволяють нам не тільки зберігати дані про багато об’єктів, але й містити кілька видів даних про кожен об’єкт.

Наприклад, у комірці нижче ми маємо два масиви. Перший, `population_amounts`, був визначений вище в розділі 2.2 і містить дані про чисельність населення світу за кожним роком (за оцінками Бюро перепису населення США). Другий масив, `years`, містить самі роки. Ці елементи впорядковані, тому рік і чисельність населення світу за цей рік мають однаковий індекс у відповідних масивах.

```python
# Просто запустіть цю клітинку

years = np.arange(1950, 2015+1)
print("Стовпець чисельності населення:", population_amounts)
print("Стовпець років:", years)
```

Припустимо, ми хочемо відповісти на таке запитання:

> У якому році чисельність населення світу перевищила 6 мільярдів?

Технічно на це запитання можна відповісти, просто вручну аналізуючи данні в масивах (див. результати виведені на екран після останньої комірки з кодом), але це дещо заплутано, оскільки доведеться підрахувати позицію, на якій чисельність населення вперше перевищила 6 мільярдів, а потім знайти відповідний елемент у масиві років. У таких випадках простіше буде помістити дані в *`Table`* — двовимірний тип набору даних.

Наведений нижче вираз:

- створює порожню таблицю за допомогою виразу `Table()`,
- додає два стовпці, викликаючи `with_columns` із чотирма аргументами (що відповідають назві стовпця, значенням для цього стовпця, назві другого стовпця та значенням для цього стовпця),
- присвоює результат імені `population`, і, нарешті,
- обчислює `population`, щоб ми могли побачити таблицю.

Строки `«Year»` та `«Population»` — це назви стовпців, які ми обрали. Імена `population_amounts` та `years` були присвоєні вище двом масивам **однакової довжини**. Функція `with_columns` (документацію можна знайти [тут](http://data8.org/fa26/reference)) приймає по черзі такі пари аргументів: рядок (що представляє назву стовпця) та масив (що представляє дані в цьому стовчику). Рядки та масиви розділяються комами. Таких пар має бути стільки скільки в таблиці буде стовпчиків.

_Примітка:_ Якщо ви хочете додати один стовпець до таблиці, ви можете використовувати або `with_columns`, або `with_column` з двома аргументами (назва стовпця та масив). Однак, якщо ви хочете додати до таблиці більше одного стовпця, ви повинні використовувати `with_columns` з чергуванням аргументів у вигляді рядків та масивів.

```python
population = Table().with_columns(
    "Population", population_amounts,
    "Year", years
)
population
```

Тепер дані об’єднано в одну таблицю! Аналізувати ці дані стало набагато простіше. Наприклад, якщо вам потрібно дізнатися, якою була чисельність населення у 1959 році, ви зможете це побачити з першого погляду.
