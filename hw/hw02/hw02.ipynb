1. *Індекс незайнятості (Non-Employment Index, NEI)*: відсоток осіб серед тих, хто здатний працювати та шукає роботу на повний робочий день, які не можуть знайти роботу.

Останню групу називають «неповна зайнятість з економічних причин», тому абревіатура цього індексу — NEI-PTER.  (Економісти чудово розбираються в маркетингу.)
